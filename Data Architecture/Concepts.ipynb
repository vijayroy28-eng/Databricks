# Data Architecture

**Data architecture** serves as a structural blueprint for an entire data solution. It is not limited to data engineering alone — it combines **analytics**, **data science**, and **data engineering** into a single design.

At its core, it defines how an organization manages data across its full lifecycle:

* **Collect** — capture data from source systems
* **Store** — persist it in appropriate storage layers
* **Integrate** — combine data across systems and domains
* **Transform** — clean, enrich, and standardize it
* **Model** — structure it for querying and analysis
* **Serve** — deliver it to consumers and applications
* **Govern** — secure, monitor, and manage its quality

Because of this breadth, data architecture is considered a major pillar of the entire data infrastructure. While the design itself is distinct from implementation — which is the role of *data engineering* — a skilled **data architect** must be deeply familiar with various technologies, frameworks, and tools to design effective, optimized solutions tailored to specific organizational needs.

## Data Maturity Levels

The creator explains **Data Maturity Levels** as a framework to categorize an organization's data readiness. There are four distinct levels that represent an organization's evolution in managing and leveraging its data:

### Level 1: Reactive

At this foundational stage, organizations lack a centralized, governed, or formal data management strategy. They function reactively, building reports, charts, or analyzing data only when specifically requested by leadership because no pre-built, centralized system exists.

### Level 2: Centralized

Organizations begin to centralize their data management practices here. They implement data warehouses or similar structures to serve as a **source of truth** for the company, allowing users to query data from a central repository rather than disparate sources.

### Level 3: Analytical

This level builds on the centralized foundation and enables deeper analytical capabilities. Organizations typically move to the cloud (e.g., *Azure*), enabling robust data governance, automated data pipelines (ETL/ELT), and comprehensive quality and security checks. The creator notes that roughly **80% of organizations** are currently in a transition phase, moving from Level 2 toward Level 3.

### Level 4: Optimization

This is the advanced stage where an organization has fully mastered the previous levels and reached a state of optimization. It involves advanced data architecture practices that focus on efficiency, quick data onboarding, and fine-tuning the entire data ecosystem.

## Architecture Design Session (ADS)

**ADS** stands for **Architecture Design Session**, which the creator describes as the fundamental **backbone** of any data architecture initiative. It is a critical, intensive process, not a quick task, and often requires a significant time investment — spanning several hours — to thoroughly address every angle and requirement of the proposed system.

### Core Characteristics of an ADS:

*   **Collaborative Environment:** The session brings together the **Data Architect**, key domain leadership (such as Finance or Analytics Directors), and one or two senior developers. Keeping the leadership group small ensures the discussion remains focused and productive.
*   **Comprehensive Discovery:** The primary purpose is to exhaustively map out the project needs. This involves a detailed questionnaire that covers **data sources**, expected **data volume**, **ingestion frequency**, and the choice between **batch processing** versus **real-time processing**.
*   **Realistic Expectations:** The creator emphasizes that a Data Architect must remain **honest and straightforward** during this session. They should not feel pressured to impress stakeholders by agreeing to unachievable requirements. Instead, it is the architect's duty to clearly identify drawbacks and technical constraints immediately.
*   **Defined Roadmap:** Once the session concludes and the design is agreed upon, it becomes the **binding plan** for the project. While stakeholders may not need to understand the deep technical specifics—such as specific cloud services or coding frameworks—they rely on the ADS to ensure their business needs are translated into a functional, realistic blueprint.


## Evolution of Data Architectures

The evolution of data architecture is a narrative of solving increasingly complex data requirements, moving from rigid relational structures to flexible, cost-effective ecosystems.

### The Traditional Era (1990s – Early 2000s)

In this phase, architectures were dominated by **Relational Data Warehouses (RDW)**, such as *Teradata*. Organizations focused on structured data, utilizing **normalization** and transactional (**OLTP**) systems to handle queries for reporting.

### The Data Lake Era (2010s)

As data variety exploded, the limitations of traditional warehouses became apparent. Organizations adopted **Data Lakes** (using storage like *ADLS Gen2* or *S3*) to store all types of data — structured, semi-structured, and unstructured — in a low-cost, scalable file system. This allowed for dumping massive amounts of raw data without immediate schema constraints.

### Modern Data Warehouse (MDW) vs. The Lakehouse (2020s)

There is a notable distinction between the *Modern Data Warehouse* and the *Lakehouse*:

* **MDW Approach:** Many organizations initially built a hybrid approach, where data was processed in a Data Lake but then moved into an RDW layer for final reporting and BI team access.
* **Lakehouse Architecture:** This is the current, most stable standard. It uses **Open Table Formats** — such as *Delta Lake*, *Apache Iceberg*, or *Apache Hudi* — directly on top of the Data Lake. This provides warehouse-level reliability (**ACID compliance**, **schema enforcement**, **time travel**) within the Data Lake environment itself, making both Data Science and BI teams efficient.


## Data Architecture Wrappers

The creator uses the analogy of a **gift** to explain the concept of **Data Architecture Wrappers**. Just as a gift requires a box, wrapping paper, and a ribbon to be presentable and usable, the core data architecture (such as a Lakehouse) requires additional structures to truly serve an organization. These wrappers are *not* the core architecture itself — they are the necessary add-ons built to meet specific business requirements and facilitate access.

### Data Marts

One primary example of a wrapper is the **Data Mart**. A Data Mart is a subset of the central data warehouse or Lakehouse, designed specifically for individual business domains such as:

* **Finance**
* **HR**
* **Marketing**

By creating these dedicated subsets, an organization:

* Prevents different teams from overwhelming the centralized warehouse with competing queries
* Ensures that each department accesses only the data relevant to their specific domain

## Data Fabric

**Data Fabric** is described as a conceptual, integrated layer built *within* an architecture rather than a specific piece of technology. Its purpose is to provide a unified way to **manage**, **govern**, and **access** data across the entire organization.

### Key Pillars of Data Fabric

| Pillar | Description |
|---|---|
| **Metadata Management** | The foundation — capturing contextual information about data (such as its location, size, and schema) to make it understandable and searchable. |
| **Data Governance and Security** | Establishing centralized policies to control who can access which data and ensuring compliance across all sources. |
| **Lineage** | Tracking the origin and movement of data, which is critical for debugging and understanding how data is transformed. |
| **Discovery** | Enabling users to easily locate the data they need within the organization's vast ecosystem. |
| **Integration and Orchestration** | Automating the flow and processing of data between various sources and destinations. |
| **Virtualization** | A powerful capability that allows users to query data from multiple sources as if it were in a single location, without needing to physically move the data. |

### Summary

Essentially, Data Fabric acts as an **enabling layer** that sits on top of all data sources and processing engines, connecting them through a centralized governance and intelligence layer.


The **Medallion Architecture** is a widely adopted standard in data engineering, rather than a strict rule, designed to create a consistent, multi-stage workflow for processing data. It organizes data into three distinct layers to manage complexity and ensure quality throughout the pipeline:

*   **Bronze Layer (Raw)**: This is the entry point where data is ingested from various sources in its original, *as-is* format. No transformations are applied here, ensuring a complete historical record.
*   **Silver Layer (Transformed)**: In this stage, data is cleaned, filtered, and transformed into a more usable format. This layer acts as the foundation for analytical processing.
*   **Gold Layer (Serving/Model)**: This is the final layer optimized for business-level reporting and analytics. Data here is modeled—often using *star* or *snowflake schemas*—to provide high performance for dashboards and downstream BI tools.

By following this layered approach, organizations establish a clear standard for data processing that is easier to maintain and troubleshoot compared to inconsistent, multi-step custom workflows.

## Lambda Architecture
![image_1795765774578.png](./image_1795765774578.png "image_1795765774578.png")

**Lambda Architecture** is a design practice in data engineering that enables organizations to leverage the strengths of both **batch** and **stream** processing within a single, unified system.

### The Core Concept

The fundamental idea behind Lambda Architecture is to process data through two parallel paths to ensure both immediate insights and comprehensive accuracy:

* **Speed Layer (Stream Processing):** This layer handles incoming data in real-time. It provides low-latency insights and updates for dashboards, which is critical for applications that require immediate decision-making or real-time user experiences.
* **Batch Layer (Slow Layer):** This layer is responsible for processing large volumes of data stored over a longer period. It provides a reliable, comprehensive **source of truth**, ensuring that historical data is accurately processed and persisted in the data lake. This layer is generally more robust but has higher latency.

| Layer | Latency | Purpose |
|---|---|---|
| **Speed Layer** | Low (near real-time) | Fast insights for dashboards and real-time decision-making |
| **Batch Layer** | High | Comprehensive, accurate historical processing and persistence |

### How It Works Together

Rather than forcing users to choose between speed and accuracy, the Lambda Architecture merges these approaches:

1. **Ingestion:** Data is ingested into a system that directs it to both the **stream processor** and the **data lake** simultaneously.
2. **Processing:** The stream processor generates quick, near real-time updates, while the batch processor runs more complex jobs on the accumulated historical data.
3. **Unified Serving:** The architecture includes an **aggregated or merged layer** at the end. Before serving data to end-users or dashboards, the system combines the results from both the speed layer and the batch layer. This ensures that the dashboards display current, real-time metrics while maintaining the consistency and accuracy of the historical data.

### Summary

This architecture is increasingly popular because it allows modern businesses to meet the demand for real-time analytics without sacrificing the deep historical analysis required for long-term reporting.

## Kappa Architecture
![image_1795765809512.png](./image_1795765809512.png "image_1795765809512.png")

The **Kappa Architecture** is a data processing design pattern that focuses on handling all data through a single **streaming layer**. Unlike the *Lambda Architecture*, which maintains separate paths for batch and streaming data, Kappa Architecture treats every piece of data as a stream, eliminating the need for a dedicated batch layer.

### Key Takeaways

*   **Unified Processing:** By utilizing a single stream-first pipeline, it aims to simplify the overall system design on paper.
*   **Complexity Trade-off:** While theoretically simpler, it can become quite complex to implement and maintain in reality, especially regarding the need to re-process historical data.
*   **Maintenance Challenges:** If a system error occurs or a manager requests a re-run of data from a specific past period, the architecture must be capable of handling that re-processing, which can be difficult in a pure streaming model.
*   **Adoption:** It is estimated that only about **5% of data solutions** leverage pure Kappa Architecture, often due to these maintenance and operational complexities.

## The Inmon Data Modeling Approach

<img src="./image_1795766499827.png" alt="Inmon architecture diagram" style="border: 4px solid #555555; border-radius: 8px; padding: 10px; max-width: 100%; box-shadow: 0 4px 10px rgba(0,0,0,0.3);" />

The **Inmon data modeling approach**, often associated with **Bill Inmon**, is fundamentally a **top-down** methodology for building data warehouses.

### Key Components of the Inmon Approach

| Component | Description |
|---|---|
| **Enterprise Data Warehouse (EDW)** | The process begins by creating a centralized, normalized Enterprise Data Warehouse. In the original Inmon methodology, this central repository is structured in the **third normal form (3NF)**, which ensures data integrity and reduces redundancy by organizing data into related tables. |
| **Top-Down Workflow** | Data is ingested from various *OLTP (Online Transactional Processing)* sources — such as databases, CRM, or ERP systems — and integrated into this single, central EDW. |
| **Creation of Data Marts** | Once the centralized EDW is established, specific **Data Marts** (e.g., for Finance or HR departments) are created as subsets of this warehouse to serve the reporting needs of different business units. |

### Important Nuance

> While the classic Inmon approach emphasizes a normalized 3NF data warehouse, many modern implementations have evolved. Data marts created from the EDW in contemporary settings are often **denormalized** to better support analytical performance, even though the core EDW remains normalized.

### Summary

By centralizing all data first, the Inmon approach aims to provide a single, consistent **"source of truth"** for the entire organization before distributing it to departmental teams.
