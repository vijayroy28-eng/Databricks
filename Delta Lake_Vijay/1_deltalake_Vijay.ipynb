# **Create Delta Table**

```python
from pyspark.sql.functions import *
```

**Using SQL API**

```python
%sql
CREATE TABLE delta_lake_vijay.default.first(

    ID INT NOT NULL,
    Salary INT
)

USING DELTA
```

**Using DELTA API**

```python
from delta.tables import *
```

```python
DeltaTable.createIfNotExists(spark)\
    .tableName("delta_lake_vijay.default.first_table_delta_api")\
        .addColumn("ID" ,"INT")\
            .addColumn("Salary", "INT")\
                .execute()
```

# Generated Columns

**Identity Columns**

```python
%sql
DROP TABLE delta_lake_vijay.default.first_table_delta_api
```

```python
DeltaTable.createOrReplace(spark)\
    .tableName("delta_lake_vijay.default.second_table_delta_api")\
        .addColumn("ID" ,"LONG",generatedAlwaysAs=IdentityGenerator())\
            .addColumn("Salary", "INT")\
                .addColumn("Name", "STRING")\
                .execute()
```

```python
%sql
Select * from delta_lake_vijay.default.second_table_delta_api
```

**Computed Columns**

```python
DeltaTable.createOrReplace(spark)\
    .tableName("delta_lake_vijay.default.second_table_delta_api")\
        .addColumn("ID" ,"LONG",generatedAlwaysAs=IdentityGenerator())\
            .addColumn("Salary", "INT")\
                .addColumn("Name", "STRING")\
                    .addColumn("SalarayAfterTax","DECIMAL(12,1)", generatedAlwaysAs="Salary*0.7")\
                .execute()
```

```python
%sql
Select * from delta_lake_vijay.default.second_table_delta_api;
```
