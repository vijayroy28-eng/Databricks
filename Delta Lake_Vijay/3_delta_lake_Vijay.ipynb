**# Schema Evolution and Schema Enforcement**

```python
data = [(100, 100000, 'John','Delhi'), (200, 200000, 'Jane','Kolkata'), (300, 300000, 'Joe','Bangalore')]

df = spark.createDataFrame(data, ['ID', 'Salary', 'Name','City'])

display(df)
```

**Use mergeSchema as True to handle Schema mismatch between source and destination**

```python
df.write.format('delta').mode('append').option('mergeSchema',True)\
.save("/Volumes/delta_lake_vijay/default/delta_volumne/demo_sink/")
```

**Reading the data dicrectly from Volumn without converting it into table**

```python
%sql
Select * from delta.`/Volumes/delta_lake_vijay/default/delta_volumne/demo_sink/`
```

# Schema Overwrite

```python
data = [(100, 100000, 'John','Delhi','HR'), (200, 200000, 'Jane','Kolkata','IT'), (300, 300000, 'Joe','Bangalore','Data Science')]

df = spark.createDataFrame(data, ['Cust_ID', 'Salary', 'Name','City','Department'])

display(df)
```

If you overwrite your data and your Source and Destination Schema is same then no ptoblem BUT if Source schema is changed and you want to OVERWRITE your data in destination then OVERWRITE the schema as well using **.option('overwriteSchema',True)**

```python
df.write.format('delta').mode('overwrite').option('overwriteSchema',True)\
.save("/Volumes/delta_lake_vijay/default/delta_volumne/demo_sink/")
```
