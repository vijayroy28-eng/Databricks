# **DML**

```python
data = [(400, 100000, 'Kaila','Delhi','HR'), (500, 200000, 'Raju','Kolkata','IT'), (600, 300000, 'Sonam','Bangalore','Data Science')]

df = spark.createDataFrame(data, ['Cust_ID', 'Salary', 'Name','City','Department'])

df.write.format('delta').mode('append')\
.save("/Volumes/delta_lake_vijay/default/delta_volumne/dml_sink/")
```

```python
%sql
Select * from delta.`/Volumes/delta_lake_vijay/default/delta_volumne/dml_sink/`
```

```python
%sql
UPDATE delta.`/Volumes/delta_lake_vijay/default/delta_volumne/dml_sink/`
SET Salary = 2 WHERE Cust_ID = 400;
```

**Deletion Vector is a Storage Optimization Technique**

Instead of rewriting an entire Parquet file or partition—which could contain thousands of records—when a specific update or deletion occurs, the system simply writes the data for the changed record and applies a deletion vector (a placeholder or flag) to the specific row in the existing file

# UPSERT

```python
data = [(400, 100000, 'Kaila','Delhi','HR'), (500, 200000, 'Raju','Kolkata','IT'), (600, 300000, 'Sonam','Bangalore','Data Science')]

df = spark.createDataFrame(data, ['Cust_ID', 'Salary', 'Name','City','Department'])

df.write.format('delta').mode('overwrite')\
.save("/Volumes/delta_lake_vijay/default/delta_volumne/dml_sink/")
```

```python
display(df)
```

Changing the NAME of CUST_ID 400 from Kaila to XX and adding 2 new CUST_IDs

```python
data = [(400, 100000, 'XX','Delhi','HR'), (4, 200000, 'Raju','Kolkata','IT'), (5, 300000, 'Sonam','Bangalore','Data Science')]

df = spark.createDataFrame(data, ['Cust_ID', 'Salary', 'Name','City','Department'])

display(df)
```

```python
from delta.tables import *
```

```python
dlt_obc = DeltaTable.forPath(spark,"/Volumes/delta_lake_vijay/default/delta_volumne/dml_sink/")
```

```python
dlt_obc.alias("trg").merge(df.alias("src"), "trg.Cust_ID == src.Cust_ID")\
    .whenMatchedUpdateAll()\
        .whenNotMatchedInsertAll()\
            .execute()
```
