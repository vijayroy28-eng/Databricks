```python
data = [(400, 100000, 'XX','Delhi','HR'), (4, 200000, 'Raju','Kolkata','IT'), (5, 300000, 'Sonam','Bangalore','Data Science')]

df = spark.createDataFrame(data, ['Cust_ID', 'Salary', 'Name','City','Department'])

display(df)
```

```python
df.write.format('delta').option('overwriteSchema', True).mode('overwrite').save('/Volumes/delta_lake_vijay/default/delta_volumne/schema_Changes/')
```
