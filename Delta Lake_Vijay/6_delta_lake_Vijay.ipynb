**CLONING**

There are two types of cloning DEEP CLONING and SHALLOW cloning.

In DEEP CLONING data is also cloned but in shallow cloning only META DATA is cloned not the DATA

```python
%sql

DESCRIBE HISTORY delta_lake_vijay.default.second_table_delta_api
```

```python
%sql
SELECT * from delta_lake_vijay.default.second_table_delta_api VERSION AS OF 3
```

```python
%sql
CREATE OR REPLACE TABLE delta_lake_vijay.default.clone_table
CLONE delta_lake_vijay.default.second_table_delta_api
```

```python
%sql
Select * from delta_lake_vijay.default.clone_table
```

```python
%sql
CREATE OR REPLACE TABLE delta_lake_vijay.default.clone_table_shallow
SHALLOW CLONE delta_lake_vijay.default.second_table_delta_api
```

```python
%sql
SELECT * FROM delta_lake_vijay.default.clone_table_shallow
```
