## 1. What are mutable and immutable objects?

**Answer:**

A **mutable** object can be modified after it is created, whereas an **immutable** object cannot be modified after creation.

### Mutable:
- `list`
- `dict`
- `set`

### Immutable:
- `int`
- `float`
- `str`
- `tuple`
- `bool`

### Example:

```python
x = [1, 2, 3]
x.append(4)       # allowed
```

**But:**

```python
x = "hello"
x[0] = "H"        # TypeError
```

## 3. Shallow copy vs Deep copy

### Shallow copy

Creates a new outer object but nested objects are still shared.

```python
import copy

a = [[1, 2], [3, 4]]
b = copy.copy(a)

b[0].append(5)

print(a)
```

**Output:**

```
[[1, 2, 5], [3, 4]]
```

### Deep copy

Copies the entire object hierarchy.

```python
b = copy.deepcopy(a)
```

Now modifying a nested object in `b` doesn't affect `a`.

## 10. Why shouldn't we use mutable objects as default arguments?

This is a very important interview question.

### Bad:

```python
def test(x=[]):
    x.append(1)
    return x

print(test())
print(test())
print(test())
```

**Output:**

```
[1]
[1, 1]
[1, 1, 1]
```

**Why?**

Because the default list is created once when the function is defined, not every time the function is called.

### Correct approach

```python
def test(x=None):
    if x is None:
        x = []
    x.append(1)
    return x
```

Now:

```python
print(test())
print(test())
```

**Output:**

```
[1]
[1]
```

## 14. What is a generator?

A generator generates values lazily, instead of storing all values in memory.

### Example:

```python
def numbers():
    for i in range(5):
        yield i
```

Then:

```python
for n in numbers():
    print(n)
```

**Why useful?**

Suppose you have 10 million records.

Instead of:

```python
return all_10_million_records
```

a generator can produce them one at a time.

**Interview answer:**

"Generators are useful for memory-efficient processing of large datasets because they produce values lazily using yield."

````python
# Alternative Approach: Single MERGE Statement for All CDC Operations
from delta.tables import DeltaTable

# This approach handles INSERT, UPDATE, DELETE in a single MERGE operation
# Assuming source has: transaction_id, amount, operation ('INSERT', 'UPDATE', 'DELETE'), sequence_id

jdbc_url = "jdbc:postgresql://host:5432/db"
target_table = "default.transactions"

# Get last processed sequence
last_seq = spark.sql(f"SELECT COALESCE(MAX(last_sequence_id), 0) FROM default.cdc_watermark WHERE table_name = 'transactions'").collect()[0][0]

# Read incremental CDC records
cdc_df = (spark.read
    .format("jdbc")
    .option("url", jdbc_url)
    .option("dbtable", f"(SELECT * FROM transactions_cdc WHERE sequence_id > {last_seq} ORDER BY sequence_id) tmp")
    .option("user", dbutils.secrets.get("postgres", "username"))
    .option("password", dbutils.secrets.get("postgres", "password"))
    .load()
)

if cdc_df.count() > 0:
    if DeltaTable.isDeltaTable(spark, target_table):
        delta_table = DeltaTable.forName(spark, target_table)
        
        # Single MERGE handling all operations
        delta_table.alias("target").merge(
            cdc_df.alias("source"),
            "target.transaction_id = source.transaction_id"
        ).whenMatchedDelete(
            condition = "source.operation = 'DELETE'"
        ).whenMatchedUpdateAll(
            condition = "source.operation = 'UPDATE'"
        ).whenNotMatchedInsert(
            condition = "source.operation = 'INSERT'",
            values = {
                "transaction_id": "source.transaction_id",
                "amount": "source.amount",
                "customer_id": "source.customer_id"
                # Add all columns except 'operation' and 'sequence_id'
            }
        ).execute()
        
        print(f"Processed {cdc_df.count()} CDC operations")
    else:
        # First load - create table with INSERT operations only
        cdc_df.filter("operation = 'INSERT'").write.format("delta").saveAsTable(target_table)
    
    # Update watermark
    new_seq = cdc_df.agg({"sequence_id": "max"}).collect()[0][0]
    spark.sql(f"""
        MERGE INTO default.cdc_watermark t
        USING (SELECT 'transactions' as table_name, {new_seq} as last_sequence_id) s
        ON t.table_name = s.table_name
        WHEN MATCHED THEN UPDATE SET t.last_sequence_id = s.last_sequence_id
        WHEN NOT MATCHED THEN INSERT *
    """)
    print(f"New watermark: {new_seq}")
````

````python

def flat_dict(data,parent_key='', sep='.'):
    items={}
    for k,v in data.items():
        if parent_key:
            new_key = f"{parent_key}{sep}{k}"
        else:
            new_key = k
        if isinstance(v,dict):
            items.update(flat_dict(v,new_key,sep=sep))
        else:
             items[new_key]=v
    return items

result = flat_dict(data)
print(result)

````

````python
# 18. Remove duplicates from a list
# Input:
data = [1, 2, 2, 3, 4, 4, 5]

# Simple solution:
# result = list(set(data))
# But this doesn't guarantee preserving original order.

# Interview-friendly solution preserving order
print("Method 1: Using list membership check")
result = []

for item in data:
    if item not in result:
        result.append(item)

print(result)

# Better for large lists (O(n) complexity):
print("\nMethod 2: Using set for faster lookup")
seen = set()
result = []

for item in data:
    if item not in seen:
        seen.add(item)
        result.append(item)

print(result)
````

````python
#Find longest Substring without repetaing characters

st = "eceba"
n = len(st)
maxi=0

for i in range(0,n):
    sub=set()
    for j in range (i,n):
        if st[j] not in sub:
            maxi=max(maxi,j-i+1)
            sub.add(st[j])
        else:
            break

print(maxi)
````

````python
#Given a string s and an integer k, find the length of the longest substring that contains at most k distinct characters.

s = "eceba"
k = 4
n= len(s)
count=0

for i in range(0,n):
    my_set=set()
    for j in range(i,n):
        my_set.add(s[j])
        if len(my_set) <=k:
            count=max(count,j-i+1)
        else:
            break

print(count)
````

````python
#"Merge Intervals"

arr = [[1,3],[2,6],[8,10],[15,18]]
n = len(arr)
arr.sort(key=lambda x:x[0])
ans = []


for i in range(0,n):
    start = arr[i][0]
    end = arr[i][1]
    if ans and start <= ans[-1][1]:
        ans[-1][1] = max(ans[-1][1],end)
    else:
        ans.append([start,end])

print(ans)
````
