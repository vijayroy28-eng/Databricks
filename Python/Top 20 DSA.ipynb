## 1. Two Sum — HashMap

> **Interview question**

Given an integer array `nums` and an integer `target`, return the indices of the two numbers such that they add up to `target`.

You may assume that each input has exactly one solution, and you cannot use the same element twice.

### Sample Input

```
nums = [2, 7, 11, 15]
target = 9
```

### Expected Output

```
[0, 1]
```

# Brute Force

````python
#Brute Force
nums = [2, 7, 11, 15]
target = 9
n= len(nums)

for i in range(0,n):
    for j in range(i+1,n):
        if nums[i] + nums[j] == target:
            print([i,j])
````

## 2. Group Anagrams — HashMap

> **Interview question**

Given an array of strings, group the strings that are anagrams of each other.

Anagrams contain exactly the same characters with the same frequencies, but potentially in a different order.

### Sample Input

```
["eat", "tea", "tan", "ate", "nat", "bat"]
```

### Expected Output

```
[ ["eat", "tea", "ate"],["tan", "nat"],["bat"] ]
```

The order of groups does not matter.

%md
# Frequency array approach
![image_1790160590368.png](./image_1790160590368.png "image_1790160590368.png")

````python
#Optimal Approach if input contains only alphbets
st = ["eat", "tea", "tan", "ate", "nat", "bat"]
n = len(st)
grp_dict = {}
for word in st:
    freq_array = [0] * 26
    for ch in word:
        ch_index = ord(ch) - ord("a")
        freq_array[ch_index] = freq_array[ch_index] + 1
    key= tuple(freq_array)
    if key not in grp_dict:
        grp_dict[key] = [word]
    else:
        grp_dict[key].append(word)
        


print(list(grp_dict.values()))
````

> **Interview question**

You are given daily sales figures. Find the maximum total sales generated over any consecutive `k` days.

### Sample Input

```
sales = [10, 20, 30, 40, 50, 60]
k = 3
```

### Expected Output

```
150
```

Because:

```
40 + 50 + 60 = 150
```

Another input

```
sales = [5, 2, 8, 1, 9, 3]
k = 2
```

Output:

```
12
```

Because:

```
8 + 1 = 9
1 + 9 = 10
9 + 3 = 12
```

### Senior DE follow-up

Suppose there are 500 million daily records. Would you load everything into a Python list?

````python
## Brute Force

sales = [100, -200, 50, 60]
k = 3
n = len(sales)
maxi=float("-inf")
for i in range (0,n-k+1):
# range(0,4-3+1) = range(0,1+1)=range(0,2)
    total=0
    for j in range(i,i+k):
# range(0,0+3)= range(0,3)
        total=total+sales[j]
    if total>maxi:
        maxi=total

print(maxi)


````

````python
## Optimal Solution - Sliding Window (O(n) time, O(1) space)

sales = [100, -200, 50, 60]
k = 3
n = len(sales)

# Step 1: Compute sum of the first window
total = sum(sales[:k])
maxi = total

# Step 2: Slide the window — subtract outgoing, add incoming
for i in range(k, n):
    total = total - sales[i - k] + sales[i]
    maxi = max(maxi, total)

print(maxi)
````

> **Interview question**

Given two strings `s` and `t`, find the smallest substring of `s` that contains all characters of `t`, including duplicate characters.

### Sample Input

```

t = "ABC"
```

### Expected Output

```
"BANC"
```

Another input

```

```

Output:

```

```

because `s` doesn't contain two `a`s.

### Follow-up

This is an excellent test of whether you understand sliding window + frequency maps.

**Optimal Approach:** https://www.youtube.com/watch?v=LxOmtM-0fss&t=2425s

![image_1791257574219.png](./image_1791257574219.png "image_1791257574219.png")

````python
s = "ADOBECODEBANC"
t = "ABC"
left = 0
right = 0
counter_target = {}
counter_window = {}
achieved = 0
ans = []
min_len = float("inf")

for ch in t:
    counter_target[ch] = counter_target.get(ch, 0) + 1
    target = len(counter_target)

for right in range(0, len(s)):
    # if s[right] in counter_target:
    counter_window[s[right]] = counter_window.get(s[right], 0) + 1
    if s[right] in counter_target and counter_window[s[right]] == counter_target[s[right]]:
        achieved += 1

    while achieved == target:  # Window is valid — keep shrinking from left
        if min_len > right - left + 1:  # if min_len is greater than curr length
            min_len = right - left + 1
            ans = [left, right]
        if s[left] in counter_target and counter_window[s[left]] == counter_target[s[left]]:
            achieved -= 1
        counter_window[s[left]] -= 1
        left = left + 1


print(s[ans[0]:ans[1]+1])
````

````python
#Valid Parenthesis
st = "{[(])}"
stk=[]

def valid_para(st="{[(])}"):
    for i in st:
        if i=="{" or i=="[" or i=="(":
            stk.append(i)
        else:
            if len(stk)==0:
                return False
            ch=stk.pop()
            if (ch=="{" and i=="}") or (ch=="[" and i=="]") or (ch=="(" and i==")"):
                continue
            else:
                return False
    return len(stk)==0

valid_para("{[()]}")


````

> **Interview question**

Given an array of integers, for every element find the first element to its right that is greater than it.

If no greater element exists, return `-1`.

### Sample Input

```
nums = [4, 5, 2, 10, 8]
```

### Expected Output

```
[5, 10, 10, -1, -1]
```

Explanation:

```
4 → 5
5 → 10
2 → 10
10 → -1
8 → -1
```

### Follow-up

Can you solve it in O(n)?

This is a good Senior DE question because it tests whether you recognize the monotonic stack pattern.

````python
#Next Greater Element — Monotonic Stack
# BRUTE FORCE

nums = [4, 5, 2, 10, 8]
n = len(nums)
ans =[-1] * n

for i in range (0,n):
    for j in range(i+1,n):
        if nums[j] > nums[i]:
            ans[i]=nums[j]
            break

print(ans)
````

## 14. Binary Search

> **Interview question**

Given a sorted array of integers and a target value, return the index of the target. If it doesn't exist, return `-1`.

### Sample Input

```
nums = [1, 3, 5, 7, 9, 11]
target = 7
```

### Expected Output

```
3
```

Another input

```
nums = [1, 3, 5, 7, 9, 11]
target = 6
```

Output:

```
-1
```

### Follow-up

Why shouldn't you use a normal linear search?

Expected discussion:

```
Linear Search → O(n)
Binary Search → O(log n)
```

**Optimal Solution:** https://www.youtube.com/watch?v=fM4Yat0dERo

![image_1791358081732.png](./image_1791358081732.png "image_1791358081732.png")

````python
nums = [1, 3, 5, 7, 9, 11]
target = 7

def binary(nums,target):
    low=0
    high=len(nums)-1
    while low <= high:
        mid=(low+high)//2
        if nums[mid]==target:
            return mid
        elif nums[mid]<target:
            low=mid+1
        else:
            high=mid-1
    return -1

binary(nums,target)

````

> **Interview question**

A sorted array has been rotated at an unknown position.

Find the index of `target` in O(log n) time.

### Sample Input

```
nums = [4, 5, 6, 7, 0, 1, 2]
target = 0
```

### Expected Output

```
4
```

Another input

```
nums = [4, 5, 6, 7, 0, 1, 2]
target = 3
```

Output:

```
-1
```

### Important follow-up

How do you determine which half of the array is sorted?

This is a very good test of binary-search understanding.

````python
nums = [4, 5, 6, 7, 0, 1, 2]
target = 5
n=len(nums)



def rot(nums,target):
    low=0
    high=n-1
    while low <= high:
        mid=(low+high)//2
        if nums[mid]==target:
            return mid
        if nums[mid] <= nums[high]:
            if nums[mid] <= target <= nums[high]:
                low = mid + 1
            else:
                high = mid - 1
        else:  # left half is sorted
            if nums[low] <= target <= nums[mid]:
                high = mid - 1
            else:
                low = mid + 1
    return -1

rot(nums,target)

````

![image_1790231397402.png](./image_1790231397402.png "image_1790231397402.png")

````python
import heapq
nums = [1,1,1,2,2,3]
k = 2
freq={}
heap=[]
res=[]
for i in range(len(nums)):
    freq[nums[i]] = freq.get(nums[i],0)+1

for key,value in freq.items():
    if len(heap) < k:
        heapq.heappush(heap,[value,key])
    else:
        heapq.heappushpop(heap,[value,key])
for j in heap:
    res.append(j[1])

print(res)

````
