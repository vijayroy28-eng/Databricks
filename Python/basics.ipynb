```python
# multi-threading is a powerful technique that allows you to process multiple tasks or operations simultaneously, rather than waiting for one to finish before starting the next
# Parallel Processing: Instead of waiting for the first item to complete, multi-threading releases resources to handle other items immediately, which is crucial when dealing with large datasets or heavy I/O operations

import time
import random
from concurrent.futures import ThreadPoolExecutor

tables = ["orders", "products","customers","reviews","cancellations"]

def my_fun(i):
    wait = 3
    time.sleep(wait)
    print(f"i am {i} took {wait} seconds")

# for i in tables:
#     my_fun(i)

with ThreadPoolExecutor(max_workers = len(tables)) as executor:
    futures = executor.map(my_fun,tables)
```
