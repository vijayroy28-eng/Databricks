In the context of this video, **SDP** stands for **Spark Declarative Pipelines**. It represents a new era of developing with *Apache Spark* and *PySpark* (1:44 - 2:08).

**What is SDP?**
*   **Declarative Framework:** Unlike traditional imperative programming where you provide step-by-step instructions for data execution, SDP allows you to define *what* you want to achieve with your data—such as transformations, filtering, or aggregations—without manually configuring the *how* (24:07 - 25:47).
*   **Automatic Optimization:** Spark handles the low-level mechanics, such as task scheduling, memory optimization, and partitioning, automatically. This eliminates the need for developers to spend time on manual performance tuning (27:49 - 28:07).
*   **Simplified Coding:** It reduces the need for complex, boilerplate code by providing higher-level APIs for common data engineering tasks, such as handling slowly changing dimensions (39:40 - 40:11).

**Key Components of SDP:**
1.  **Flows:** The foundational unit of data processing in SDP, which handles reading from a source, applying transformations, and writing to a target (41:00 - 42:24).
2.  **Data Sets:** These are the queryable output objects of your flows, such as **Streaming Tables**, **Materialized Views**, and **Temporary Views** (45:12 - 46:16).
3.  **DAGs:** SDP automatically constructs a Directed Acyclic Graph (DAG) based on the dependencies defined in your code, so you do not need to build it manually (51:34 - 52:24).

Essentially, SDP is designed to make data engineering more accessible and efficient by letting developers focus on business logic while the framework manages the underlying execution complexity (33:06 - 33:55).

In the context of **Spark Declarative Pipelines (SDP)**, a **Data Set** is a queryable object that serves as the output of a flow within a pipeline. When you define a flow to process data—reading from a source, applying transformations, and writing to a destination—the resulting object is considered a data set (45:11-45:52).

There are three primary types of data sets available in SDP, each serving different processing needs:

*   **Streaming Table:** This is a managed table designed as a **streaming target** (47:44-48:06). It is built on top of *Spark Structured Streaming*, which guarantees **idempotency** (exactly-once processing) (204:18-204:57). It is best suited for incremental, append-only data ingestion where you want to process only new data (209:29-209:34).
*   **Materialized View:** These are managed tables used for **batch targets** (48:45-49:10). Unlike standard views that recompute every time, materialized views cache the results of a query and refresh them on an interval (131:16-131:38). They handle incremental updates efficiently and are ideal when you need to perform complex aggregations (49:12-49:43).
*   **Temporary View:** This is a standard, non-persisted view that **holds only a query** rather than storing physical data (49:45-50:10). It is useful for creating intermediate, transient steps within your pipeline that do not require storing results in the lake (155:38-155:55).
