# Delta Sharing

The creator introduces **Delta Sharing** (2:20:40) as a solution for sharing data across different *Databricks workspaces* that are not connected to the same *Unity Catalog metastore*.

### Key Concepts of Delta Sharing:
* **Bridging Metastores:** While data typically stays within a single metastore, Delta Sharing allows for secure, cross-workspace data access between different regions or environments (2:21:43).
* **Databricks-to-Databricks Sharing:** The video focuses on using this feature to share catalogs seamlessly between distinct Databricks environments (2:22:37).

### Implementation Steps described in the video:
1. **Define a Recipient:** You must first create a **recipient** in your sharing workspace by using the unique sharing identifier from the destination workspace (2:23:57 - 2:24:28).
2. **Grant Permissions:** To ensure the sharing works, specific administrative permissions must be granted at the **metastore level** in the source workspace. This includes `USE PROVIDER`, `USE RECIPIENT`, and `USE SHARE` permissions (2:26:31).
3. **Result:** Once configured, the destination user can query the shared catalog as if it were local, while the actual data remains securely in the source's catalog (2:27:02).
