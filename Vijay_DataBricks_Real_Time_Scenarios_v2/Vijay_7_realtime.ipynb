In the context of *Azure Databricks* and *Unity Catalog*, a **User Defined Function (UDF)** allows you to create custom logic that can be stored and reused across your entire data environment, rather than existing only within the scope of a single notebook or cluster (2:28:16 - 2:28:41).

### Key Concepts of Catalog-Level UDFs:
* **Persistence:** Unlike standard PySpark UDFs that vanish when a session or cluster terminates, these UDFs are registered in the *Unity Catalog*. This means they remain available for future use cases, even after clusters are restarted or projects are promoted to new environments (2:36:16 - 2:36:40).
* **Language Support:** You can create these functions using either **SQL** or **Python**, providing flexibility depending on your development requirements (2:30:57 - 2:31:09).
* **Structure:** They are typically scalar functions, meaning they return a single value (e.g., string, integer, float) based on the input provided (2:30:33 - 2:30:50).

### How to Create a UDF:
1. **Define the Function:** You use a standard SQL `CREATE OR REPLACE FUNCTION` command. You must define the function name (following the three-level namespace: *catalog.schema.function*), parameters, return type, and the logic itself (2:29:25 - 2:30:30).
2. **Logic Implementation:**
    * **SQL:** You write the transformation directly within the function block (2:31:13 - 2:31:40).
    * **Python:** You specify the language as `PYTHON` and wrap your Python code within dollar signs (e.g., `$$ return ... $$`) (2:34:57 - 2:35:43).
3. **Application:** Once created, these functions can be called just like built-in SQL functions within your queries (e.g., `SELECT function_name(column) FROM table`), making your data transformations cleaner and more modular (2:34:08 - 2:34:40).
