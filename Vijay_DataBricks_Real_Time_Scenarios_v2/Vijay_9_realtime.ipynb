# DELTA LIVE TABLE - DLT

## Create Streaming table

```python
import dlt

from pyspark.sql.functions import *
```

**Creating a Table**

**Expectations**

```python
rule = {
        "rule1" : "ID is NOT NULL",
        "rule2" : "Salary > 10"
    }
```

```python
@dlt.table

@dlt.expect_all_or_drop(rule)

def stream_stg():
    df = spark.readStream.table('real_time_catalog.source.my_table')
    return df
```

**Creating View on TOP of my source table**

```python
@dlt.view

def steam_view():

    df = spark.readStream.table('stream_stg')

    df.withColumn('flag',lit("Hi"))

    return df
```

**Streaming Table Final**

```python
@dlt.table

def stream_final():
    df = spark.readStream.table('steam_view')

    return df
```
