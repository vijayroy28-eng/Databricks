# How to ingest only newly arrived files

```python
from pyspark.sql.functions import *
from pyspark.sql.window import *
from pyspark.sql.types import *
from delta.tables import *
```

```python
orders_schema ="""
order_id INT,
customer_id INT,
order_date DATE,
amount DECIMAL(10,2)
"""
```

```python
df = spark.readStream.format('csv').option('header', True)\
    .schema(orders_schema)\
        .option('path','/Volumes/pyspark_cata_vijay/source/pyspark_volume_practice/orders/checkPoint')\
        .load('/Volumes/pyspark_cata_vijay/source/pyspark_volume_practice/orders/')
```

```python
df.writeStream.format('delta')\
    .option("checkpointLocation","/Volumes/pyspark_cata_vijay/source/pyspark_volume_practice/orders/checkPoint")\
        .option('mergeSchema', True)\
            .trigger(once=True)\
                .start('/Volumes/pyspark_cata_vijay/source/pyspark_volume_practice/orders/')
```
