```python
from pyspark.sql.functions import *
from pyspark.sql.window import *
from pyspark.sql.types import *
from delta.tables import *
```

# Read Data from JSON

```python
df = spark.read.format('json')\
    .option('inferSchema',True)\
        .option('multiLine',True)\
        .load('/Volumes/pyspark_cata_vijay/source/pyspark_volume_practice/apidata.json')


display(df)
```

```python
df = df.select("customer.*","customer.location.*","*","payment.*")\
    .withColumn("delivery_updates",explode("delivery_updates"))\
        .withColumn("items",explode("items"))\
            .drop("location").drop("customer").drop("payment")


# display(df)
```

```python
df = df.select("items.*","*")

display(df)
```
