```python
from pyspark.sql.functions import *
from pyspark.sql.window import *
from pyspark.sql.types import *
from delta.tables import *
```

```python
class DataValidation:
    def __init__(self,df):
        self.df = df

    def dedup(self,KeyCol,OrderCol):
        df = self.df.withColumn("rn", row_number().over(Window.partitionBy(KeyCol).orderBy(desc(OrderCol))))
        df = df.filter(col("rn") == 1).drop("rn")
        return df
```

```python
data = 
[
("1","abc@gmail.com","abc","NY","USA","yes","1","2022-01-01","100","USD","COD"),
("1","abc@gmail.com","abc","NY","USA","yes","2","2022-01-01","100","USD","COD"),
("2","def@gmail.com","def","LA","USA","yes","1","2022-01-02","200","USD","COD"),
("3","ghi@gmail.com","ghi","LA","USA","yes","1","2022-01-03","300","USD","COD"),
("4","jkl@gmail.com","jkl","LA","USA","yes","1","2022-01-04","400","USD","COD")
]

columns = ["customer_id","email","name","city","country","delivery_updates","order_id","order_timestamp","amount","currency","method"]

df = spark.createDataFrame(data,columns)

display(df)
```

```python
objj = DataValidation(df)
```

```python
df_dedup = objj.dedup("customer_id","order_timestamp")

display(df_dedup)
```
